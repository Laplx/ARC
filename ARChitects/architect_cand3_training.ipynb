```python
import os
print(os.getcwd())
os.chdir("/data/coding/ARC")

from dataclasses import dataclass
import os

@dataclass
class Config:
    data_root: str = 'data'
    split: str = 'training'
    max_tasks: int = 0
    model_key: str = ''
    model_path: str = 'outputs/arc_lora_sft_C/lora_merged_model'
    tokenizer_path: str = 'outputs/arc_lora_sft_C/lora_merged_model'
    max_new_tokens: int = 931
    output_dir: str = 'outputs/architect_eval'
    viz_failures: bool = False

cfg = Config()

def get_truth(task, context):
    tests = task.get('test') or []
    if len(tests) != 1:
        return None
    return tests[0].get('output')

def get_task_id(task, context):
    return task.get('task_id', context.get('index'))
```

```python
from ARChitects.architect_transampling_candall import build_architect_solver

solver = build_architect_solver(
    model_key=cfg.model_key,
    model_path=cfg.model_path,
    tokenizer_path=cfg.tokenizer_path,
    max_new_tokens=cfg.max_new_tokens,
)
```

```python
from eval.core_cand import ARCDataset, run_evaluation
dataset = ARCDataset(root=cfg.data_root, split=cfg.split, max_tasks=0, start_from=0)

reports = run_evaluation(
    dataset=dataset,
    solver=solver,
    output_dir=cfg.output_dir,
    model_id=cfg.model_key,
    model_key='archi_training',
    viz_failures=cfg.viz_failures,
    get_truth=get_truth,
    get_task_id=get_task_id,
    pass_k=8,
)
print(reports['summary'])
```
