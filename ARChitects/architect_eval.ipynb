# Architect Evaluation

Minimal eval for NVARC-style Architect DFS

```python
from dataclasses import dataclass
import os

@dataclass
class Config:
    data_root: str = 'data'
    split: str = 'evaluation'
    max_tasks: int = 0
    model_key: str = ''
    model_path: str = ''
    tokenizer_path: str = ''
    max_new_tokens: int = 512
    output_dir: str = 'outputs/architect_eval'
    viz_failures: bool = True

cfg = Config()
```

```python
from ARChitects.architect_dfs import build_architect_solver
from eval.core import ARCDataset, run_evaluation

solver = build_architect_solver(
    model_key=cfg.model_key,
    model_path=cfg.model_path or None,
    tokenizer_path=cfg.tokenizer_path or None,
    max_new_tokens=cfg.max_new_tokens,
)
dataset = ARCDataset(root=cfg.data_root, split=cfg.split, max_tasks=cfg.max_tasks)

reports = run_evaluation(
    dataset=dataset,
    solver=solver,
    output_dir=cfg.output_dir,
    model_id=cfg.model_key,
    model_key='architect_dfs',
    viz_failures=cfg.viz_failures,
)
print(reports['summary'])
```
